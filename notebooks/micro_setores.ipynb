```python
import pandas as pd
import numpy as np

def dissecar_setores_multi_sessoes(arquivos_sessoes, carros_alvo=[10, 80]):
    print("Dissecação Multi-Sessões (Full Time Sports)")
    print("=" * 85)

    for nome_sessao, caminho_csv in arquivos_sessoes.items():
        print(f"\nPROCESSANDO: {nome_sessao}")
        print("-" * 85)
        
        try:
            df = pd.read_csv(caminho_csv, sep=';', decimal=',') 
            df['Carro'] = df['Piloto'].astype(str).str.extract(r'^(\d+)').astype(float)
            
            colunas_alvo = ['S1 Tm', 'S2 Tm', 'S3 Tm', 'SPT']
            for col in colunas_alvo:
                if df[col].dtype == object:
                    df[col] = df[col].astype(str).str.replace(',', '.')
                df[col] = pd.to_numeric(df[col], errors='coerce')
                df.loc[df[col] <= 5.0, col] = np.nan 
                
            melhores_setores_geral = df[['S1 Tm', 'S2 Tm', 'S3 Tm']].min()
            resultados = []
            
            for carro in carros_alvo:
                df_carro = df[df['Carro'] == carro]
                
                if not df_carro.empty:
                    best_s1 = df_carro['S1 Tm'].min()
                    best_s2 = df_carro['S2 Tm'].min()
                    best_s3 = df_carro['S3 Tm'].min()
                    best_spt = df_carro['SPT'].max()
                    
                    ideal_lap = best_s1 + best_s2 + best_s3
                    
                    resultados.append({
                        'Carro': int(carro),
                        'Melhor_S1 (s)': round(best_s1, 3),
                        'Melhor_S2 (s)': round(best_s2, 3),
                        'Melhor_S3 (s)': round(best_s3, 3),
                        'Speed_Trap (km/h)': round(best_spt, 1),
                        'Volta_Ideal (s)': round(ideal_lap, 3)
                    })

            if resultados:
                df_setores = pd.DataFrame(resultados)
                print("PERFORMANCE ABSOLUTA DOS PILOTOS:")
                print(df_setores.to_string(index=False))
                
                print("\nREFERÊNCIA (O MELHOR DA PISTA NESTA SESSÃO):")
                print(f"S1: {round(melhores_setores_geral['S1 Tm'], 3)}s | "
                      f"S2: {round(melhores_setores_geral['S2 Tm'], 3)}s | "
                      f"S3: {round(melhores_setores_geral['S3 Tm'], 3)}s")
            else:
                print("Nenhum dado válido encontrado para os carros alvo nesta sessão.")
                
        except Exception as e:
            print(f"Erro ao processar o arquivo {caminho_csv}: {e}")
            
    print("=" * 85)

# --- ÁREA DE EXECUÇÃO ---
# Dicionário corrigido com a nomenclatura exata que você pediu:
sessoes_curvelo = {
    'Treino (T2)': '../data/02_interim/dados_limpos_T2.csv',
    'Corrida (P1)': '../data/02_interim/dados_limpos_P1.csv'
}

dissecar_setores_multi_sessoes(sessoes_curvelo, carros_alvo=[10, 80])
```
