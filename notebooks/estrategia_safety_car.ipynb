```python
import pandas as pd

def calcular_delta_safety_car_pre_corrida(piloto, ritmo_treino, perda_box_estimada, reducao_sc_percentual):
    print(f"MATRIZ DE RISCO PRÉ-CORRIDA (SAFETY CAR): {piloto}")
    print("-" * 75)
    
    # Tempo absoluto = Ritmo base do TL2 + O que esperamos perder no Pit Lane a 60km/h
    tempo_absoluto_pit = ritmo_treino + perda_box_estimada
    resultados = []
    
    for reducao in reducao_sc_percentual:
        # Calcula como o pelotão vai rastejar na pista sob SC
        ritmo_sc = ritmo_treino * (1 + (reducao / 100))
        
        # O novo custo real do nosso pit stop
        perda_box_sc = tempo_absoluto_pit - ritmo_sc
        vantagem_sc = perda_box_estimada - perda_box_sc
        
        resultados.append({
            'Redução SC (%)': f"-{reducao}%",
            'Ritmo Pelotão SC (s)': round(ritmo_sc, 3),
            'Custo Pit SC (s)': round(perda_box_sc, 3),
            'Tempo Salvo (s)': round(vantagem_sc, 3)
        })
        
    df_sc = pd.DataFrame(resultados)
    print(df_sc.to_string(index=False))
    print(f"Resumo: Planejamos uma perda normal de {perda_box_estimada}s em bandeira verde.")
    print("-" * 75 + "\n")

# --- ÁREA DE EXECUÇÃO ---

# 1. ZONTA: Ritmo base extraído do TL2
calcular_delta_safety_car_pre_corrida(
    piloto="Ricardo Zonta (Carro 10)",
    ritmo_treino=84.867,       # O ritmo real do Long Run dele no Treino
    perda_box_estimada=33.200, # A estimativa padrão de Pit Stop da equipe
    reducao_sc_percentual=[25, 30, 35, 40]
)

# 2. IBIAPINA: Ritmo base extraído do TL2
calcular_delta_safety_car_pre_corrida(
    piloto="Alfredinho Ibiapina (Carro 80)",
    ritmo_treino=84.256,       # O ritmo real do Long Run dele no Treino
    perda_box_estimada=33.200, # A equipe planeja a mesma operação no box
    reducao_sc_percentual=[25, 30, 35, 40]
)
```
