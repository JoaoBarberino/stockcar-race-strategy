```python
import pandas as pd

def fundir_telemetria_e_pit(csv_voltas, csv_pitstops, csv_final):
    print("Fundindo Voltas e Pit Stops")

    # 1. Carrega as duas bases que você já limpou
    df_voltas = pd.read_csv(csv_voltas, sep=';', decimal=',')
    df_pitstops = pd.read_csv(csv_pitstops, sep=';', decimal=',')

    # 2. Cria a "Chave" de conexão: O número do carro
    df_voltas['Carro'] = df_voltas['Piloto'].str.extract(r'^(\d+)').astype(str)
    
    # Garante que as chaves de busca (Carro e Lap) sejam lidas como texto (string) para não dar erro de formato
    df_pitstops['Carro'] = df_pitstops['Carro'].astype(str)
    df_voltas['Lap'] = df_voltas['Lap'].astype(str).str.replace(r'\.0$', '', regex=True)
    df_pitstops['Lap'] = df_pitstops['Lap'].astype(str).str.replace(r'\.0$', '', regex=True)

    # 3. Left Join
    # Procura o Carro e a Volta do pitstop e cola na linha exata correspondente da planilha de voltas
    df_final = pd.merge(df_voltas, df_pitstops[['Carro', 'Lap', 'Pit Time']], on=['Carro', 'Lap'], how='left')

    # 4. Reorganiza para ficar bonito de ler
    colunas = df_final.columns.tolist()
    # Move o 'Carro' para a primeira coluna
    colunas.insert(0, colunas.pop(colunas.index('Carro')))
    # Move o 'Pit Time' para perto do Tempo de Volta
    colunas.insert(colunas.index('Lap Tm (Segundos)') + 1, colunas.pop(colunas.index('Pit Time')))
    df_final = df_final[colunas]

    # 5. Salva a Base Definitiva
    df_final.to_csv(csv_final, index=False, sep=';', decimal=',')
    print(f"Base de dados definitiva montada com sucesso em: {csv_final}")

# --- ÁREA DE EXECUÇÃO ---
arquivo_voltas = '../data/02_interim/dados_limpos_P1.csv'
arquivo_pitstops = '../data/02_interim/dados_pitstops_P1.csv'
arquivo_definitivo = '../data/03_processed/TELEMETRIA_FINAL_P1.csv'

# Rode a fusão
fundir_telemetria_e_pit(arquivo_voltas, arquivo_pitstops, arquivo_definitivo)
```
